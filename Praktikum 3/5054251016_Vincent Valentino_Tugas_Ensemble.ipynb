```python
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, confusion_matrix, ConfusionMatrixDisplay, roc_auc_score
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from imblearn.over_sampling import SMOTE, RandomOverSampler
from imblearn.under_sampling import RandomUnderSampler
from imblearn.pipeline import Pipeline
```

```python
credit = credit.drop_duplicates()
x = credit.drop(columns='Class')
y = credit['Class']
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, stratify=y, random_state=42)
```

```python
samplers = {
    'No Sampling': None,
    'Random Oversampling': RandomOverSampler(random_state=42),
    'Random Undersampling': RandomUnderSampler(random_state=42),
    'SMOTE': SMOTE(random_state=42)
}
models = {
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    'AdaBoost' : AdaBoostClassifier(n_estimators=100, random_state=42)
}
```

```python
results = []
confusion = {}
for model_name, model in models.items():
    for sampler_name, sampler in samplers.items():
        if sampler is None:
            model.fit(x_train, y_train)
            y_pred = model.predict(x_test)
            y_prob = model.predict_proba(x_test)[:, 1]

        else:
            pipeline = Pipeline([
                ('sampler', sampler),
                ('model', model)
            ])

            pipeline.fit(x_train, y_train)

            y_pred = pipeline.predict(x_test)
            y_prob = pipeline.predict_proba(x_test)[:, 1]

        results.append({
            'Model': model_name,
            'Sampling': sampler_name,
            'Accuracy': accuracy_score(y_test, y_pred),
            'Precision': precision_score(y_test, y_pred),
            'Recall': recall_score(y_test, y_pred),
            'F1': f1_score(y_test, y_pred),
            'ROC-AUC': roc_auc_score(y_test, y_prob)
        })

        confusion[(model_name, sampler_name)] = confusion_matrix(y_test, y_pred)
```

```python
results_df = pd.DataFrame(results)
results_df
```

```python
fig, axes = plt.subplots(2, 4,figsize=(16, 8))
axes = axes.flatten()
for ax, ((model_name, sampler_name), cm) in zip(axes, confusion.items()):
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Normal', 'Fraud'])
    disp.plot(ax=ax, colorbar=False)
    ax.set_title(f'{model_name} {sampler_name}')

plt.tight_layout()
plt.savefig('confusion.png',dpi=300)
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(30, 5))
axes = axes.flatten()
for ax, score in zip(axes, ['F1', 'Recall', 'Precision']):
    sb.barplot(data=results_df, x='Sampling', y=score, hue='Model', ax=ax)
    ax.set_title(f'{score}: Sampling and Ensemble Comparison')
plt.tight_layout()
plt.savefig('metric.png',dpi=300)
plt.show()
```
